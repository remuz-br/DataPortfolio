```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
tips_dataset = sns.load_dataset('tips')
```

```python
print(tips_dataset.head(10))
print(tips_dataset.info(10))
print(tips_dataset.describe())
```

```python
print(tips_dataset.isna())
print(tips_dataset.isna().sum())
```

```python
sns.histplot(tips_dataset['total_bill'], bins=20, kde=True, color='skyblue')
plt.show()
sns.histplot(tips_dataset['tip'], bins=20, kde=True, color='pink')
plt.show()
```

```python
sns.scatterplot(tips_dataset, x='total_bill', y='tip', hue='sex')
plt.show()
print(tips_dataset['total_bill'].corr(tips_dataset['tip']))
```

```python
sns.scatterplot(tips_dataset, x='total_bill', y='tip', hue='smoker')
plt.show()
```

```python
sns.regplot(tips_dataset, x='total_bill', y='tip')
```

```python
sns.barplot(tips_dataset,x='tip', y='day', hue='time')
plt.show()
```

```python
tip_percent = (tips_dataset["tip"] / tips_dataset['total_bill']) * 100
```

```python
sns.histplot(tip_percent, bins=20, kde=True, color='green')
plt.xlabel('percentage')
plt.show()
```

```python
sns.scatterplot(x=tips_dataset['total_bill'], y=tip_percent, color='blue')
plt.show()
print('Correlation between tips percentage from bill and total bill: ', tips_dataset['total_bill'].corr(tip_percent))
```

```python
sns.pairplot(tips_dataset, hue='sex', kind='reg')
plt.show()
```

```python
sns.FacetGrid(tips_dataset, col='smoker' ).map_dataframe(sns.regplot, x='total_bill', y='tip')
plt.show()
```
