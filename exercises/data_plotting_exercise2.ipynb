```python

g = sns.FacetGrid(tips_dataset, col="smoker", hue="sex", height=4, aspect=1)
g.map_dataframe(sns.scatterplot, x="total_bill", y="tip")
g.add_legend()
g.set_axis_labels("Total Bill ($)", "Tip ($)")
g.set_titles("Smoker: {col_name}")
g.fig.suptitle("Tip vs Total Bill — Grouped by Smoker and Gender", y=1.05)

plt.show()
```
