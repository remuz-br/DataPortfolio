```python
from collections import Counter
titanic_embark_count = Counter(titanic_dataset['embark_town'])
destin = []
dst_cnt = []
for destination, count in titanic_embark_count.items():
    destin.append(destination)
    dst_cnt.append(count)
print(destin)
print(dst_cnt)
```

```python
plt.pie(dst_cnt, labels=destin, autopct='%1.1f%%', startangle=90)
plt.show()
```
